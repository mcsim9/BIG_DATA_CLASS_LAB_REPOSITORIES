```python
<!DOCTYPE html>
<html lang="fr">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>TOP 18 Flagships — AI Use Case Registry</title>
<link rel="preconnect" href="https://fonts.googleapis.com">
<link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
<link href="https://fonts.googleapis.com/css2?family=Fraunces:opsz,wght@9..144,400;9..144,500;9..144,600;9..144,700&family=IBM+Plex+Sans:wght@300;400;500;600;700&family=IBM+Plex+Mono:wght@400;500&display=swap" rel="stylesheet">
<style>
  :root {
    --ink:        #0a1628;
    --ink-2:      #1e2a3f;
    --ink-soft:   #4a5568;
    --muted:      #8896a8;
    --line:       #e4e8ee;
    --line-2:     #eef1f5;
    --paper:      #fbfbf8;
    --paper-2:    #ffffff;
    --accent:     #8b1e2d;     /* deep burgundy — institutional */
    --accent-2:   #c89b3c;     /* aged gold accent */
    --ok:         #1f6b4a;
    --warn:       #b4691f;
    --shadow:     0 1px 0 rgba(10,22,40,.04), 0 8px 24px -12px rgba(10,22,40,.12);
    --shadow-lg:  0 1px 0 rgba(10,22,40,.04), 0 24px 48px -20px rgba(10,22,40,.18);
    --radius:     2px;
    --radius-lg:  4px;
  }

  * { box-sizing: border-box; margin: 0; padding: 0; }

  html, body {
    background: var(--paper);
    color: var(--ink);
    font-family: 'IBM Plex Sans', sans-serif;
    font-size: 14px;
    line-height: 1.5;
    -webkit-font-smoothing: antialiased;
    min-height: 100vh;
  }

  body {
    background-image:
      radial-gradient(at 0% 0%, rgba(139,30,45,.025) 0%, transparent 50%),
      radial-gradient(at 100% 100%, rgba(200,155,60,.02) 0%, transparent 50%);
  }

  /* ============ TOP BAR ============ */
  .topbar {
    border-bottom: 1px solid var(--line);
    background: var(--paper-2);
    position: sticky; top: 0; z-index: 100;
    backdrop-filter: blur(8px);
  }
  .topbar-inner {
    max-width: 1400px; margin: 0 auto;
    padding: 18px 32px;
    display: flex; align-items: center; justify-content: space-between;
    gap: 24px;
  }
  .brand {
    display: flex; align-items: center; gap: 14px;
  }
  .brand-mark {
    width: 36px; height: 36px;
    background: var(--ink);
    color: var(--paper);
    display: grid; place-items: center;
    font-family: 'Fraunces', serif;
    font-weight: 600;
    font-size: 18px;
    letter-spacing: -.02em;
    position: relative;
  }
  .brand-mark::after {
    content: ""; position: absolute; inset: 3px;
    border: 1px solid var(--accent-2); opacity: .5;
  }
  .brand-text {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: 18px;
    letter-spacing: -.01em;
  }
  .brand-text small {
    display: block;
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    color: var(--muted);
    letter-spacing: .12em;
    text-transform: uppercase;
    margin-top: 2px;
  }
  .topbar-actions {
    display: flex; align-items: center; gap: 12px;
  }
  .meta-tag {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 11px;
    color: var(--muted);
    letter-spacing: .08em;
    text-transform: uppercase;
  }

  /* ============ BREADCRUMB ============ */
  .breadcrumb {
    max-width: 1400px; margin: 0 auto;
    padding: 20px 32px 8px;
    display: flex; align-items: center; gap: 10px;
    font-family: 'IBM Plex Mono', monospace;
    font-size: 11px;
    letter-spacing: .1em;
    text-transform: uppercase;
    color: var(--muted);
  }
  .breadcrumb a {
    color: var(--muted);
    text-decoration: none;
    transition: color .2s;
    cursor: pointer;
  }
  .breadcrumb a:hover { color: var(--accent); }
  .breadcrumb .sep { opacity: .4; }
  .breadcrumb .current { color: var(--ink); }

  /* ============ MAIN ============ */
  main {
    max-width: 1400px; margin: 0 auto;
    padding: 16px 32px 80px;
  }

  /* ============ PAGE HEADER ============ */
  .page-header {
    margin-bottom: 32px;
    padding-bottom: 24px;
    border-bottom: 1px solid var(--line);
    display: flex; justify-content: space-between; align-items: flex-end;
    gap: 32px;
  }
  .page-header h1 {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: clamp(32px, 4vw, 48px);
    letter-spacing: -.02em;
    line-height: 1.05;
    color: var(--ink);
  }
  .page-header h1 em {
    font-style: italic;
    color: var(--accent);
    font-weight: 400;
  }
  .page-header .sub {
    color: var(--ink-soft);
    margin-top: 12px;
    max-width: 600px;
    font-size: 15px;
  }
  .page-header-meta {
    text-align: right;
    flex-shrink: 0;
  }
  .page-header-meta .num {
    font-family: 'Fraunces', serif;
    font-size: 56px;
    font-weight: 400;
    line-height: 1;
    color: var(--ink);
    letter-spacing: -.04em;
  }
  .page-header-meta .label {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    letter-spacing: .15em;
    text-transform: uppercase;
    color: var(--muted);
    margin-top: 6px;
  }

  /* ============ TOOLBAR ============ */
  .toolbar {
    display: flex; gap: 12px; align-items: center;
    margin-bottom: 24px;
    flex-wrap: wrap;
  }
  .search {
    flex: 1; min-width: 260px;
    position: relative;
  }
  .search input {
    width: 100%;
    padding: 10px 14px 10px 36px;
    border: 1px solid var(--line);
    background: var(--paper-2);
    font-family: 'IBM Plex Sans', sans-serif;
    font-size: 13px;
    color: var(--ink);
    border-radius: var(--radius);
    transition: border-color .2s;
  }
  .search input:focus { outline: none; border-color: var(--ink); }
  .search::before {
    content: "";
    position: absolute; left: 12px; top: 50%; transform: translateY(-50%);
    width: 14px; height: 14px;
    border: 1.5px solid var(--muted);
    border-radius: 50%;
  }
  .search::after {
    content: "";
    position: absolute; left: 22px; top: calc(50% + 4px);
    width: 6px; height: 1.5px;
    background: var(--muted);
    transform: rotate(45deg);
  }

  .btn {
    padding: 10px 16px;
    border: 1px solid var(--ink);
    background: var(--ink);
    color: var(--paper);
    font-family: 'IBM Plex Sans', sans-serif;
    font-size: 12px;
    font-weight: 500;
    letter-spacing: .04em;
    text-transform: uppercase;
    cursor: pointer;
    border-radius: var(--radius);
    transition: all .15s;
    display: inline-flex; align-items: center; gap: 8px;
  }
  .btn:hover { background: var(--accent); border-color: var(--accent); }
  .btn-ghost {
    background: transparent; color: var(--ink); border-color: var(--line);
  }
  .btn-ghost:hover { background: var(--ink); color: var(--paper); border-color: var(--ink); }
  .btn-sm { padding: 7px 12px; font-size: 11px; }

  /* ============ FLAGSHIP TABLE (TOP 18) ============ */
  .table-wrap {
    background: var(--paper-2);
    border: 1px solid var(--line);
    border-radius: var(--radius-lg);
    overflow: hidden;
    box-shadow: var(--shadow);
  }
  table.flagships {
    width: 100%;
    border-collapse: collapse;
    font-size: 13px;
  }
  table.flagships thead th {
    text-align: left;
    padding: 14px 16px;
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    font-weight: 500;
    letter-spacing: .12em;
    text-transform: uppercase;
    color: var(--muted);
    border-bottom: 1px solid var(--line);
    background: linear-gradient(180deg, #fafaf6 0%, #f6f6f1 100%);
    white-space: nowrap;
  }
  table.flagships tbody tr {
    border-bottom: 1px solid var(--line-2);
    cursor: pointer;
    transition: background .15s;
  }
  table.flagships tbody tr:hover {
    background: #fdfaf3;
  }
  table.flagships tbody tr:last-child { border-bottom: none; }
  table.flagships tbody td {
    padding: 16px;
    vertical-align: top;
  }
  .rank {
    font-family: 'Fraunces', serif;
    font-size: 22px;
    font-weight: 500;
    color: var(--ink);
    letter-spacing: -.02em;
    width: 56px;
  }
  .rank-1, .rank-2, .rank-3 { color: var(--accent); }
  .flag-name {
    font-weight: 600;
    color: var(--ink);
    font-size: 14px;
  }
  .flag-desc {
    color: var(--ink-soft);
    font-size: 12px;
    margin-top: 4px;
    max-width: 420px;
    line-height: 1.4;
  }
  .pill {
    display: inline-block;
    padding: 3px 8px;
    border: 1px solid var(--line);
    border-radius: 2px;
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    letter-spacing: .08em;
    text-transform: uppercase;
    color: var(--ink-soft);
    background: var(--paper);
    white-space: nowrap;
  }
  .pill-accent {
    border-color: var(--accent);
    color: var(--accent);
    background: rgba(139,30,45,.04);
  }
  .pill-gold {
    border-color: var(--accent-2);
    color: #8a6a1f;
    background: rgba(200,155,60,.06);
  }
  .uc-count {
    font-family: 'Fraunces', serif;
    font-size: 18px;
    color: var(--ink);
  }
  .uc-count small {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    color: var(--muted);
    letter-spacing: .1em;
    text-transform: uppercase;
    margin-left: 4px;
  }

  /* ============ FLAGSHIP DETAIL ============ */
  .flagship-header {
    display: grid;
    grid-template-columns: 1fr 320px;
    gap: 48px;
    margin-bottom: 40px;
    padding-bottom: 32px;
    border-bottom: 1px solid var(--line);
  }
  .flag-rank-big {
    font-family: 'Fraunces', serif;
    font-style: italic;
    font-size: 14px;
    color: var(--accent);
    letter-spacing: .08em;
    text-transform: uppercase;
    margin-bottom: 8px;
  }
  .flag-title {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: clamp(28px, 3.5vw, 42px);
    letter-spacing: -.02em;
    line-height: 1.1;
    color: var(--ink);
    margin-bottom: 16px;
  }
  .flag-description {
    color: var(--ink-soft);
    font-size: 15px;
    line-height: 1.6;
    max-width: 640px;
  }

  .info-card {
    background: var(--paper-2);
    border: 1px solid var(--line);
    border-radius: var(--radius-lg);
    padding: 20px;
    box-shadow: var(--shadow);
  }
  .info-card h3 {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    font-weight: 500;
    letter-spacing: .15em;
    text-transform: uppercase;
    color: var(--muted);
    margin-bottom: 16px;
    padding-bottom: 10px;
    border-bottom: 1px solid var(--line-2);
  }
  .info-row {
    display: grid;
    grid-template-columns: 110px 1fr;
    gap: 12px;
    padding: 8px 0;
    font-size: 12px;
    align-items: baseline;
  }
  .info-row .key {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    letter-spacing: .08em;
    text-transform: uppercase;
    color: var(--muted);
  }
  .info-row .val {
    color: var(--ink);
    font-weight: 500;
  }

  /* Section title */
  .section-title {
    display: flex; align-items: baseline; justify-content: space-between;
    margin-bottom: 20px;
    padding-bottom: 12px;
    border-bottom: 1px solid var(--line);
  }
  .section-title h2 {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: 24px;
    letter-spacing: -.01em;
    color: var(--ink);
  }
  .section-title h2 em {
    color: var(--muted);
    font-style: italic;
    font-size: 18px;
    margin-left: 8px;
    font-weight: 400;
  }

  /* Use case cards */
  .uc-grid {
    display: grid;
    grid-template-columns: repeat(auto-fill, minmax(320px, 1fr));
    gap: 16px;
  }
  .uc-card {
    background: var(--paper-2);
    border: 1px solid var(--line);
    border-radius: var(--radius-lg);
    padding: 20px;
    cursor: pointer;
    transition: all .2s;
    position: relative;
    box-shadow: var(--shadow);
  }
  .uc-card:hover {
    border-color: var(--ink);
    transform: translateY(-2px);
    box-shadow: var(--shadow-lg);
  }
  .uc-card.new {
    border: 1px dashed var(--muted);
    background: transparent;
    box-shadow: none;
    display: flex; align-items: center; justify-content: center;
    min-height: 180px;
    color: var(--muted);
    font-family: 'IBM Plex Mono', monospace;
    font-size: 11px;
    letter-spacing: .12em;
    text-transform: uppercase;
  }
  .uc-card.new:hover { border-color: var(--ink); color: var(--ink); }
  .uc-card .uc-name {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: 18px;
    color: var(--ink);
    letter-spacing: -.01em;
    margin-bottom: 8px;
    line-height: 1.25;
  }
  .uc-card .uc-desc {
    color: var(--ink-soft);
    font-size: 12px;
    line-height: 1.5;
    margin-bottom: 16px;
  }
  .uc-card .uc-meta {
    display: flex; flex-wrap: wrap; gap: 6px;
    padding-top: 12px;
    border-top: 1px solid var(--line-2);
  }
  .uc-card .countries-strip {
    display: flex; gap: 4px; margin-top: 12px;
    flex-wrap: wrap;
  }
  .country-flag {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    font-weight: 500;
    color: var(--ink);
    padding: 2px 6px;
    background: var(--paper);
    border: 1px solid var(--line);
    border-radius: 2px;
    letter-spacing: .04em;
  }

  /* ============ USE CASE DETAIL ============ */
  .uc-detail-grid {
    display: grid;
    grid-template-columns: 1fr 1fr;
    gap: 16px;
    margin-bottom: 32px;
  }
  .field {
    display: flex; flex-direction: column;
    gap: 4px;
  }
  .field label {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    letter-spacing: .12em;
    text-transform: uppercase;
    color: var(--muted);
    font-weight: 500;
  }
  .field input, .field select, .field textarea {
    padding: 9px 12px;
    border: 1px solid var(--line);
    background: var(--paper-2);
    font-family: 'IBM Plex Sans', sans-serif;
    font-size: 13px;
    color: var(--ink);
    border-radius: var(--radius);
    transition: border-color .15s;
  }
  .field input:focus, .field select:focus, .field textarea:focus {
    outline: none; border-color: var(--ink);
  }
  .field textarea { resize: vertical; min-height: 72px; }
  .field.span-2 { grid-column: span 2; }
  .field.readonly input, .field.readonly textarea {
    background: var(--paper);
    color: var(--ink);
    border-color: transparent;
    cursor: default;
    font-weight: 500;
  }

  /* ============ COUNTRIES TABLE ============ */
  .country-chips {
    display: flex; gap: 8px; flex-wrap: wrap;
    margin-bottom: 24px;
  }
  .country-chip {
    padding: 10px 16px;
    border: 1px solid var(--line);
    background: var(--paper-2);
    border-radius: var(--radius);
    cursor: pointer;
    font-size: 12px;
    font-weight: 500;
    transition: all .15s;
    display: flex; align-items: center; gap: 8px;
  }
  .country-chip:hover { border-color: var(--ink); }
  .country-chip.active {
    background: var(--ink);
    color: var(--paper);
    border-color: var(--ink);
  }
  .country-chip .code {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 10px;
    opacity: .7;
    letter-spacing: .1em;
  }
  .country-chip.add {
    border-style: dashed;
    color: var(--muted);
  }
  .country-chip.add:hover { color: var(--ink); border-style: solid; }

  /* ============ VALUE BREAKDOWN ============ */
  .value-grid {
    display: grid;
    grid-template-columns: repeat(auto-fit, minmax(180px, 1fr));
    gap: 1px;
    background: var(--line);
    border: 1px solid var(--line);
    border-radius: var(--radius-lg);
    overflow: hidden;
    margin-bottom: 24px;
  }
  .value-cell {
    background: var(--paper-2);
    padding: 16px;
  }
  .value-cell .vlabel {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 9px;
    letter-spacing: .12em;
    text-transform: uppercase;
    color: var(--muted);
    margin-bottom: 6px;
    line-height: 1.3;
  }
  .value-cell .vnum {
    font-family: 'Fraunces', serif;
    font-size: 22px;
    font-weight: 500;
    color: var(--ink);
    letter-spacing: -.01em;
  }
  .value-cell .vnum input {
    border: none;
    background: transparent;
    font-family: 'Fraunces', serif;
    font-size: 22px;
    font-weight: 500;
    color: var(--ink);
    width: 100%;
    padding: 0;
  }
  .value-cell .vnum input:focus { outline: none; border-bottom: 1px solid var(--ink); }
  .value-cell .vunit {
    font-family: 'IBM Plex Mono', monospace;
    font-size: 11px;
    color: var(--muted);
    margin-left: 4px;
  }
  .value-cell.total {
    background: var(--ink);
    color: var(--paper);
  }
  .value-cell.total .vlabel { color: rgba(255,255,255,.6); }
  .value-cell.total .vnum, .value-cell.total .vnum input {
    color: var(--paper);
  }

  /* ============ MODAL ============ */
  .modal-overlay {
    position: fixed; inset: 0;
    background: rgba(10,22,40,.4);
    backdrop-filter: blur(4px);
    display: none;
    align-items: center; justify-content: center;
    z-index: 1000;
    padding: 32px;
  }
  .modal-overlay.active { display: flex; }
  .modal {
    background: var(--paper-2);
    border-radius: var(--radius-lg);
    width: 100%; max-width: 600px;
    max-height: 90vh;
    overflow-y: auto;
    box-shadow: var(--shadow-lg);
    border: 1px solid var(--line);
  }
  .modal-header {
    padding: 24px 28px;
    border-bottom: 1px solid var(--line);
    display: flex; justify-content: space-between; align-items: center;
  }
  .modal-header h3 {
    font-family: 'Fraunces', serif;
    font-weight: 500;
    font-size: 20px;
    letter-spacing: -.01em;
  }
  .modal-close {
    background: none; border: none; font-size: 24px;
    cursor: pointer; color: var(--muted); line-height: 1;
  }
  .modal-close:hover { color: var(--ink); }
  .modal-body { padding: 24px 28px; }
  .modal-footer {
    padding: 16px 28px;
    border-top: 1px solid var(--line);
    display: flex; gap: 12px; justify-content: flex-end;
    background: var(--paper);
  }

  /* status dots */
  .status-dot {
    display: inline-block; width: 6px; height: 6px;
    border-radius: 50%;
    margin-right: 6px;
    vertical-align: middle;
  }
  .status-dot.prod { background: var(--ok); }
  .status-dot.pilot { background: var(--warn); }
  .status-dot.dev { background: var(--muted); }

  /* helpers */
  .hidden { display: none !important; }
  .row-flex { display: flex; gap: 12px; align-items: center; flex-wrap: wrap; }
  .spacer { flex: 1; }

  /* fade-in transitions */
  .view { animation: fadeIn .3s ease; }
  @keyframes fadeIn {
    from { opacity: 0; transform: translateY(4px); }
    to { opacity: 1; transform: translateY(0); }
  }

  /* JSON export */
  .json-output {
    background: var(--ink);
    color: #c8d4e0;
    padding: 16px;
    border-radius: var(--radius);
    font-family: 'IBM Plex Mono', monospace;
    font-size: 11px;
    line-height: 1.6;
    max-height: 400px;
    overflow: auto;
    white-space: pre;
  }

  /* responsive */
  @media (max-width: 900px) {
    .flagship-header { grid-template-columns: 1fr; }
    .uc-detail-grid { grid-template-columns: 1fr; }
    .field.span-2 { grid-column: span 1; }
    .topbar-inner { padding: 14px 20px; }
    main { padding: 16px 20px 60px; }
    .breadcrumb { padding: 16px 20px 4px; }
  }

</style>
</head>
<body>

<header class="topbar">
  <div class="topbar-inner">
    <div class="brand">
      <div class="brand-mark">F</div>
      <div class="brand-text">
        Flagship Registry
        <small>AI Use Case Portfolio · v1.0</small>
      </div>
    </div>
    <div class="topbar-actions">
      <span class="meta-tag" id="last-updated">Updated · 18.05.2026</span>
      <button class="btn btn-ghost btn-sm" onclick="openExport()">Export JSON</button>
    </div>
  </div>
</header>

<nav class="breadcrumb" id="breadcrumb"></nav>

<main id="app"></main>

<!-- Modal -->
<div class="modal-overlay" id="modal">
  <div class="modal">
    <div class="modal-header">
      <h3 id="modal-title">—</h3>
      <button class="modal-close" onclick="closeModal()">×</button>
    </div>
    <div class="modal-body" id="modal-body"></div>
    <div class="modal-footer" id="modal-footer"></div>
  </div>
</div>

<script>
/* ===========================================================
   STATE — mock data, 18 flagships, several use cases each
   =========================================================== */

const STATE = {
  view: 'top18',        // top18 | flagship | usecase | country
  flagshipId: null,
  usecaseId: null,
  country: null,
  search: '',
};

const POLES = ['Retail Banking', 'Corporate & Investment', 'Asset Management', 'Insurance', 'Group Functions'];
const ENTITIES = ['BNP Paribas SA', 'Cardif', 'Arval', 'BNL', 'BancWest', 'Consorsbank'];
const AI_ACT_CATS = ['Minimal Risk', 'Limited Risk', 'High Risk', 'Unacceptable'];
const TECHS = ['LLM (GPT-4o)', 'LLM (Claude 3.5)', 'Random Forest', 'XGBoost', 'CNN', 'Transformer', 'BERT', 'Rule-based + ML'];
const PLATFORMS = ['Azure ML', 'GCP Vertex', 'Internal — Echo', 'AWS SageMaker', 'Databricks'];
const BRICKS = ['NLP', 'Computer Vision', 'Tabular ML', 'GenAI', 'Speech', 'Forecasting'];
const DOMAINS = ['Risk', 'Compliance', 'Customer Experience', 'Operations', 'Fraud Detection', 'Trading'];
const CATEGORIES = ['Predictive', 'Generative', 'Classification', 'Recommendation', 'Anomaly Detection'];
const SOURCING = ['In-house', 'Vendor (OpenAI)', 'Vendor (Anthropic)', 'Hybrid', 'Open-source'];
const COUNTRIES_LIST = [
  { code: 'FR', name: 'France' },
  { code: 'IT', name: 'Italy' },
  { code: 'BE', name: 'Belgium' },
  { code: 'LU', name: 'Luxembourg' },
  { code: 'DE', name: 'Germany' },
  { code: 'ES', name: 'Spain' },
  { code: 'PL', name: 'Poland' },
  { code: 'US', name: 'United States' },
  { code: 'UK', name: 'United Kingdom' },
  { code: 'NL', name: 'Netherlands' },
];
const STATUSES = ['Production', 'Pilot', 'Development', 'On Hold', 'Decommissioned'];
const MEASURE_TYPES = ['Estimated', 'Measured', 'Validated'];
const ROLLOUT = ['Local', 'Regional', 'Global', 'Multi-country'];
const MRR_CATS = ['Critical', 'Core', 'Supporting', 'Exploratory'];

/* 18 Flagship use cases — realistic banking AI portfolio */
const FLAGSHIPS_SEED = [
  { rank: 1, name: 'Athena — Credit Decisioning Suite', desc: 'End-to-end retail credit decisioning with explainable AI scoring across 11 countries.', sponsor: 'C. Mercier', owner: 'L. Tran', division: 'Retail Banking', subdiv: 'Consumer Credit', poc: 'athena-poc@grp.com', analyst: 'M. Boiral' },
  { rank: 2, name: 'Sentinel — Transaction Fraud Engine', desc: 'Real-time fraud detection over card and SEPA payment flows with adaptive thresholds.', sponsor: 'A. Dupont', owner: 'P. Rossi', division: 'Group Risk', subdiv: 'Fraud & Financial Crime', poc: 'sentinel@grp.com', analyst: 'K. Nguyen' },
  { rank: 3, name: 'Helios — KYC Document Intelligence', desc: 'Automated document classification and data extraction for onboarding workflows.', sponsor: 'S. Klein', owner: 'M. Garcia', division: 'Compliance', subdiv: 'Financial Security', poc: 'helios-kyc@grp.com', analyst: 'F. Achille' },
  { rank: 4, name: 'Atlas — Wealth Advisory Copilot', desc: 'Generative AI assistant for private bankers — portfolio summarization and client-ready notes.', sponsor: 'V. Bianchi', owner: 'H. Lefèvre', division: 'Asset Management', subdiv: 'Private Banking', poc: 'atlas@grp.com', analyst: 'D. Petit' },
  { rank: 5, name: 'Iris — Customer Voice Analytics', desc: 'Speech-to-insight on call center recordings for satisfaction, intent and compliance monitoring.', sponsor: 'M. Owens', owner: 'R. Schmidt', division: 'Retail Banking', subdiv: 'Customer Experience', poc: 'iris-voice@grp.com', analyst: 'J. Maréchal' },
  { rank: 6, name: 'Forge — Code Generation Platform', desc: 'Internal developer copilot with private-model deployment and code intelligence on legacy COBOL.', sponsor: 'T. Bauer', owner: 'N. Costa', division: 'Group IT', subdiv: 'Engineering Productivity', poc: 'forge-dev@grp.com', analyst: 'L. Roux' },
  { rank: 7, name: 'Orion — AML Behavior Monitoring', desc: 'Graph-based anomaly detection over correspondent banking flows for sanctions screening.', sponsor: 'A. Dupont', owner: 'I. Kowalski', division: 'Compliance', subdiv: 'AML', poc: 'orion-aml@grp.com', analyst: 'B. Hassan' },
  { rank: 8, name: 'Pulse — Market News Synthesizer', desc: 'Real-time financial news summarization and event tagging for trading desks.', sponsor: 'E. Morel', owner: 'F. Albrecht', division: 'Corporate & Investment', subdiv: 'Global Markets', poc: 'pulse-news@grp.com', analyst: 'S. Bernard' },
  { rank: 9, name: 'Mercury — Claims Triage', desc: 'Insurance claims classification and severity scoring with image analysis for auto and home.', sponsor: 'C. Fontaine', owner: 'V. Russo', division: 'Insurance', subdiv: 'Cardif Non-Life', poc: 'mercury@grp.com', analyst: 'O. Bensaïd' },
  { rank: 10, name: 'Vega — ESG Risk Scoring', desc: 'Climate and ESG transition risk scoring on corporate counterparts using NLP on reports.', sponsor: 'J. Andersson', owner: 'A. Mazet', division: 'Group Risk', subdiv: 'Sustainability', poc: 'vega-esg@grp.com', analyst: 'P. Vidal' },
  { rank: 11, name: 'Nova — Branch Demand Forecast', desc: 'Branch traffic and cash demand forecasting to optimize staffing and logistics.', sponsor: 'C. Mercier', owner: 'D. Lambert', division: 'Retail Banking', subdiv: 'Network Operations', poc: 'nova@grp.com', analyst: 'T. Marchand' },
  { rank: 12, name: 'Echo — Conversational Banking', desc: 'Customer-facing chatbot across web and mobile, multilingual, with handoff to human agents.', sponsor: 'M. Owens', owner: 'B. Caron', division: 'Retail Banking', subdiv: 'Digital Channels', poc: 'echo-bot@grp.com', analyst: 'R. Lemaire' },
  { rank: 13, name: 'Talos — Cybersecurity Detection', desc: 'Anomaly detection over network and endpoint telemetry — XDR-grade autonomous triage.', sponsor: 'P. Singh', owner: 'M. Eriksson', division: 'Group IT', subdiv: 'Cybersecurity', poc: 'talos@grp.com', analyst: 'C. Diallo' },
  { rank: 14, name: 'Lumen — Loan Origination Assistant', desc: 'Document understanding for SME loan files — pre-fills application data and flags inconsistencies.', sponsor: 'V. Bianchi', owner: 'G. Moreau', division: 'Corporate & Investment', subdiv: 'SME Lending', poc: 'lumen@grp.com', analyst: 'N. Said' },
  { rank: 15, name: 'Quasar — Liquidity Stress Simulator', desc: 'ML-based simulation of liquidity shocks under regulatory and idiosyncratic scenarios.', sponsor: 'E. Morel', owner: 'S. Chen', division: 'Group Risk', subdiv: 'ALM', poc: 'quasar@grp.com', analyst: 'A. Pirelli' },
  { rank: 16, name: 'Tempo — HR Talent Matching', desc: 'Internal mobility — semantic matching of skills to open roles across group entities.', sponsor: 'F. Laurent', owner: 'J. Patel', division: 'Group Functions', subdiv: 'Human Resources', poc: 'tempo@grp.com', analyst: 'E. Fontaine' },
  { rank: 17, name: 'Argos — Contract Intelligence', desc: 'Legal contract clause extraction and comparison for trade finance and ISDA agreements.', sponsor: 'S. Klein', owner: 'L. Bach', division: 'Legal', subdiv: 'Contract Management', poc: 'argos-legal@grp.com', analyst: 'V. Kapoor' },
  { rank: 18, name: 'Cypher — Card Acquisition Targeting', desc: 'Propensity modeling for premium card cross-sell, with uplift modeling on direct campaigns.', sponsor: 'C. Mercier', owner: 'O. Bouchard', division: 'Retail Banking', subdiv: 'Marketing', poc: 'cypher@grp.com', analyst: 'R. Khan' },
];

/* Generate use cases per flagship */
function randPick(arr) { return arr[Math.floor(Math.random() * arr.length)]; }
function randInt(min, max) { return Math.floor(Math.random() * (max - min + 1)) + min; }

const USE_CASE_TEMPLATES = [
  ['Real-time scoring API', 'Streaming inference endpoint integrated with core systems'],
  ['Batch nightly run', 'End-of-day batch scoring for regulatory reporting'],
  ['Operator dashboard', 'Internal cockpit with case management and review queues'],
  ['Customer-facing flow', 'Embedded in self-service onboarding journey'],
  ['Cross-border extension', 'Replication of the core engine for new geographies'],
  ['Champion-challenger', 'A/B framework for continuous model improvement'],
  ['Regulatory reporting feed', 'Output streamed into compliance reporting toolchain'],
];

function generateUseCases(flagship) {
  const n = randInt(2, 4);
  const picks = USE_CASE_TEMPLATES.slice().sort(() => Math.random() - 0.5).slice(0, n);
  return picks.map((tpl, i) => {
    const countriesCount = randInt(1, 4);
    const countries = COUNTRIES_LIST.slice().sort(() => Math.random() - 0.5).slice(0, countriesCount);
    return {
      id: `${flagship.rank}-${i + 1}`,
      name: `${tpl[0]} — ${flagship.name.split('—')[0].trim()}`,
      description: tpl[1] + '.',
      pole: randPick(POLES),
      entity: randPick(ENTITIES),
      aiActCategory: randPick(AI_ACT_CATS),
      technology: randPick(TECHS),
      platform: randPick(PLATFORMS),
      mainBrick: randPick(BRICKS),
      aiDomain: randPick(DOMAINS),
      aiCategory: randPick(CATEGORIES),
      sourcing: randPick(SOURCING),
      iaValueTotal: randInt(150, 8500),
      modelContributors: ['Data Science Lab', 'MLOps Squad ' + flagship.entity.split(' ')[0]].join(', '),
      pilotCountry: countries[0].name,
      itsvc2: 'ITSVC-' + randInt(2024, 2026) + '-' + randInt(100, 999),
      dateItsvc: `2025-${String(randInt(1,12)).padStart(2,'0')}-${String(randInt(1,28)).padStart(2,'0')}`,
      lastGate: ['Concept', 'Pilot Gate', 'Industrialization Gate', 'Production Gate'][randInt(0,3)],
      dateLastGate: `2025-${String(randInt(1,12)).padStart(2,'0')}-${String(randInt(1,28)).padStart(2,'0')}`,
      countries: countries.map(c => ({
        code: c.code,
        name: c.name,
        year: 2025,
        status: randPick(STATUSES),
        productionDate: `2025-${String(randInt(1,12)).padStart(2,'0')}-${String(randInt(1,28)).padStart(2,'0')}`,
        valueCapped: randInt(50, 2500),
        netBanking: randInt(20, 800),
        grossCostSavings: randInt(30, 900),
        costAvoidance: randInt(10, 400),
        costOfRisk: randInt(0, 350),
        rwaSavings: randInt(0, 250),
        valueMeasureType: randPick(MEASURE_TYPES),
        rollout: randPick(ROLLOUT),
        mrrCategory: randPick(MRR_CATS),
        mrr: randInt(15, 220),
        useCaseOwner: flagship.owner,
      })),
    };
  });
}

const FLAGSHIPS = FLAGSHIPS_SEED.map(f => ({
  ...f,
  useCases: generateUseCases(f),
}));

/* ===========================================================
   ROUTING / RENDERING
   =========================================================== */

const app = document.getElementById('app');
const breadcrumb = document.getElementById('breadcrumb');

function navigate(view, opts = {}) {
  STATE.view = view;
  if (opts.flagshipId !== undefined) STATE.flagshipId = opts.flagshipId;
  if (opts.usecaseId !== undefined) STATE.usecaseId = opts.usecaseId;
  if (opts.country !== undefined) STATE.country = opts.country;
  render();
  window.scrollTo({ top: 0, behavior: 'smooth' });
}

function render() {
  renderBreadcrumb();
  if (STATE.view === 'top18') renderTop18();
  else if (STATE.view === 'flagship') renderFlagship();
  else if (STATE.view === 'usecase') renderUseCase();
  else if (STATE.view === 'country') renderCountry();
}

function renderBreadcrumb() {
  const parts = [`<a onclick="navigate('top18')">TOP 18</a>`];
  if (STATE.view !== 'top18') {
    const f = FLAGSHIPS.find(x => x.rank === STATE.flagshipId);
    if (f) parts.push(`<span class="sep">/</span><a onclick="navigate('flagship', {flagshipId:${f.rank}})">${escapeHtml(f.name.split('—')[0].trim())}</a>`);
  }
  if (STATE.view === 'usecase' || STATE.view === 'country') {
    const f = FLAGSHIPS.find(x => x.rank === STATE.flagshipId);
    const uc = f?.useCases.find(u => u.id === STATE.usecaseId);
    if (uc) parts.push(`<span class="sep">/</span><a onclick="navigate('usecase', {usecaseId:'${uc.id}'})">${escapeHtml(truncate(uc.name, 40))}</a>`);
  }
  if (STATE.view === 'country') {
    parts.push(`<span class="sep">/</span><span class="current">${STATE.country}</span>`);
  }
  // Mark last as current
  if (parts.length > 0) {
    const lastIdx = parts.length - 1;
    if (!parts[lastIdx].includes('current')) {
      parts[lastIdx] = parts[lastIdx].replace('<a', '<span class="current"').replace('</a>', '</span>');
    }
  }
  breadcrumb.innerHTML = parts.join('');
}

/* ----- TOP 18 view ----- */
function renderTop18() {
  const q = STATE.search.toLowerCase();
  const rows = FLAGSHIPS
    .filter(f => !q || f.name.toLowerCase().includes(q) || f.desc.toLowerCase().includes(q) || f.division.toLowerCase().includes(q))
    .map(f => `
      <tr onclick="navigate('flagship', {flagshipId:${f.rank}})">
        <td class="rank rank-${f.rank <= 3 ? f.rank : 'n'}">${String(f.rank).padStart(2, '0')}</td>
        <td>
          <div class="flag-name">${escapeHtml(f.name)}</div>
          <div class="flag-desc">${escapeHtml(f.desc)}</div>
        </td>
        <td><span class="pill">${escapeHtml(f.division)}</span></td>
        <td><span class="pill pill-gold">${escapeHtml(f.subdiv)}</span></td>
        <td>${escapeHtml(f.owner)}<br><small style="color:var(--muted);font-size:11px">${escapeHtml(f.poc)}</small></td>
        <td><span class="uc-count">${f.useCases.length}<small>cases</small></span></td>
      </tr>
    `).join('');

  app.innerHTML = `
    <div class="view">
      <div class="page-header">
        <div>
          <h1>The <em>TOP 18</em> Flagships</h1>
          <p class="sub">Strategic AI initiatives across the group — ranked by impact, scope and maturity. Each flagship anchors one or more downstream use cases deployed across markets.</p>
        </div>
        <div class="page-header-meta">
          <div class="num">18</div>
          <div class="label">Active flagships</div>
        </div>
      </div>

      <div class="toolbar">
        <div class="search">
          <input type="text" placeholder="Search flagship, division, business analyst…" value="${escapeHtml(STATE.search)}" oninput="STATE.search=this.value; renderTop18();">
        </div>
        <button class="btn btn-ghost" onclick="openExport()">Export JSON</button>
      </div>

      <div class="table-wrap">
        <table class="flagships">
          <thead>
            <tr>
              <th>Rank</th>
              <th>Flagship</th>
              <th>Division</th>
              <th>Sub-division</th>
              <th>Use Case Owner</th>
              <th>Use Cases</th>
            </tr>
          </thead>
          <tbody>${rows}</tbody>
        </table>
      </div>
    </div>
  `;
}

/* ----- FLAGSHIP DETAIL view ----- */
function renderFlagship() {
  const f = FLAGSHIPS.find(x => x.rank === STATE.flagshipId);
  if (!f) { navigate('top18'); return; }

  const ucCards = f.useCases.map(uc => `
    <div class="uc-card" onclick="navigate('usecase', {usecaseId:'${uc.id}'})">
      <div class="uc-name">${escapeHtml(uc.name)}</div>
      <div class="uc-desc">${escapeHtml(uc.description)}</div>
      <div class="uc-meta">
        <span class="pill pill-accent">${escapeHtml(uc.aiCategory)}</span>
        <span class="pill">${escapeHtml(uc.aiDomain)}</span>
      </div>
      <div class="countries-strip">
        ${uc.countries.map(c => `<span class="country-flag">${c.code}</span>`).join('')}
      </div>
    </div>
  `).join('');

  app.innerHTML = `
    <div class="view">
      <div class="flagship-header">
        <div>
          <div class="flag-rank-big">Flagship n° ${String(f.rank).padStart(2, '0')} of 18</div>
          <h1 class="flag-title">${escapeHtml(f.name)}</h1>
          <p class="flag-description">${escapeHtml(f.desc)}</p>
        </div>
        <div class="info-card">
          <h3>Flagship characteristics</h3>
          <div class="info-row"><span class="key">Sponsor</span><span class="val">${escapeHtml(f.sponsor)}</span></div>
          <div class="info-row"><span class="key">Use Case Owner</span><span class="val">${escapeHtml(f.owner)}</span></div>
          <div class="info-row"><span class="key">Division</span><span class="val">${escapeHtml(f.division)}</span></div>
          <div class="info-row"><span class="key">Sub-division</span><span class="val">${escapeHtml(f.subdiv)}</span></div>
          <div class="info-row"><span class="key">Point of Contact</span><span class="val">${escapeHtml(f.poc)}</span></div>
          <div class="info-row"><span class="key">AI Business Analyst</span><span class="val">${escapeHtml(f.analyst)}</span></div>
        </div>
      </div>

      <div class="section-title">
        <h2>Use cases <em>${f.useCases.length}</em></h2>
        <button class="btn btn-sm" onclick="openCreateUseCase(${f.rank})">+ New use case</button>
      </div>

      <div class="uc-grid">
        ${ucCards}
        <div class="uc-card new" onclick="openCreateUseCase(${f.rank})">
          + Create new use case
        </div>
      </div>
    </div>
  `;
}

/* ----- USE CASE DETAIL view ----- */
function renderUseCase() {
  const f = FLAGSHIPS.find(x => x.rank === STATE.flagshipId);
  const uc = f?.useCases.find(u => u.id === STATE.usecaseId);
  if (!uc) { navigate('flagship', {flagshipId: STATE.flagshipId}); return; }

  const countryChips = uc.countries.map(c => `
    <div class="country-chip" onclick="navigate('country', {country:'${c.code}'})">
      <span class="code">${c.code}</span>
      <span>${escapeHtml(c.name)}</span>
    </div>
  `).join('');

  app.innerHTML = `
    <div class="view">
      <div class="page-header">
        <div>
          <div class="flag-rank-big" style="color:var(--muted)">Use case · ${escapeHtml(f.name.split('—')[0].trim())}</div>
          <h1>${escapeHtml(uc.name)}</h1>
          <p class="sub">${escapeHtml(uc.description)}</p>
        </div>
        <div class="page-header-meta">
          <div class="num">${uc.iaValueTotal.toLocaleString('fr-FR')}</div>
          <div class="label">IA Value Total · k€</div>
        </div>
      </div>

      <div class="section-title">
        <h2>Characteristics</h2>
        <button class="btn btn-sm btn-ghost" onclick="alert('Mode édition à brancher.')">Edit</button>
      </div>

      <div class="uc-detail-grid">
        ${fieldRO('Name', uc.name, true)}
        ${fieldRO('Pole', uc.pole)}
        ${fieldRO('Entity', uc.entity)}
        ${fieldRO('AI Act Category', uc.aiActCategory)}
        ${fieldRO('Technology', uc.technology)}
        ${fieldRO('Platform', uc.platform)}
        ${fieldRO('Main AI Brick', uc.mainBrick)}
        ${fieldRO('AI Domain', uc.aiDomain)}
        ${fieldRO('AI Category', uc.aiCategory)}
        ${fieldRO('Sourcing', uc.sourcing)}
        ${fieldRO('Pilot Country', uc.pilotCountry)}
        ${fieldRO('ITSVC 2', uc.itsvc2)}
        ${fieldRO('Date ITSVC', uc.dateItsvc)}
        ${fieldRO('Last Gate Passed', uc.lastGate)}
        ${fieldRO('Date Last Gate Passed', uc.dateLastGate)}
        ${fieldRO('Model Development Contributors', uc.modelContributors, true)}
      </div>

      <div class="section-title">
        <h2>Deployment by country <em>${uc.countries.length}</em></h2>
        <button class="btn btn-sm" onclick="openAddCountry('${uc.id}')">+ Add country</button>
      </div>

      <p style="color:var(--ink-soft); font-size:13px; margin-bottom:16px;">Click a country to drill down into year, status, AI value breakdown and MRR by territory.</p>

      <div class="country-chips">
        ${countryChips}
        <div class="country-chip add" onclick="openAddCountry('${uc.id}')">+ Add country</div>
      </div>
    </div>
  `;
}

/* ----- COUNTRY DETAIL view ----- */
function renderCountry() {
  const f = FLAGSHIPS.find(x => x.rank === STATE.flagshipId);
  const uc = f?.useCases.find(u => u.id === STATE.usecaseId);
  const c = uc?.countries.find(x => x.code === STATE.country);
  if (!c) { navigate('usecase', {usecaseId: STATE.usecaseId}); return; }

  const statusClass = c.status === 'Production' ? 'prod' : c.status === 'Pilot' ? 'pilot' : 'dev';

  app.innerHTML = `
    <div class="view">
      <div class="page-header">
        <div>
          <div class="flag-rank-big" style="color:var(--muted)">${escapeHtml(uc.name)}</div>
          <h1>${escapeHtml(c.name)} <em style="font-size:.6em; font-family:'IBM Plex Mono', monospace; color:var(--muted); letter-spacing:.08em;">${c.code}</em></h1>
          <p class="sub">Deployment metrics and value attribution for the ${c.code} market.</p>
        </div>
        <div class="page-header-meta">
          <div class="num">${c.valueCapped.toLocaleString('fr-FR')}</div>
          <div class="label">AI value capped · k€</div>
        </div>
      </div>

      <div class="section-title">
        <h2>Deployment</h2>
      </div>

      <div class="uc-detail-grid" style="margin-bottom:32px;">
        ${fieldEdit('Year', c.year, 'number', `updateCountry('${uc.id}','${c.code}','year',this.value)`)}
        ${fieldSelect('Status', c.status, STATUSES, `updateCountry('${uc.id}','${c.code}','status',this.value)`)}
        ${fieldEdit('Production date', c.productionDate, 'date', `updateCountry('${uc.id}','${c.code}','productionDate',this.value)`)}
        ${fieldSelect('AI Value Measure Type', c.valueMeasureType, MEASURE_TYPES, `updateCountry('${uc.id}','${c.code}','valueMeasureType',this.value)`)}
        ${fieldSelect('Rollout', c.rollout, ROLLOUT, `updateCountry('${uc.id}','${c.code}','rollout',this.value)`)}
        ${fieldSelect('MRR use case category', c.mrrCategory, MRR_CATS, `updateCountry('${uc.id}','${c.code}','mrrCategory',this.value)`)}
        ${fieldEdit('MRR', c.mrr, 'number', `updateCountry('${uc.id}','${c.code}','mrr',this.value)`)}
        ${fieldRO('AI Use Case Owner', c.useCaseOwner)}
      </div>

      <div class="section-title">
        <h2>AI Value — full year <em>k€</em></h2>
      </div>

      <div class="value-grid">
        <div class="value-cell total">
          <div class="vlabel">Total · capped</div>
          <div class="vnum">${c.valueCapped.toLocaleString('fr-FR')}<span class="vunit">k€</span></div>
        </div>
        <div class="value-cell">
          <div class="vlabel">Net Banking Impact</div>
          <div class="vnum"><input type="number" value="${c.netBanking}" onchange="updateCountry('${uc.id}','${c.code}','netBanking',this.value)"><span class="vunit">k€</span></div>
        </div>
        <div class="value-cell">
          <div class="vlabel">Gross cost savings</div>
          <div class="vnum"><input type="number" value="${c.grossCostSavings}" onchange="updateCountry('${uc.id}','${c.code}','grossCostSavings',this.value)"><span class="vunit">k€</span></div>
        </div>
        <div class="value-cell">
          <div class="vlabel">Cost avoidance</div>
          <div class="vnum"><input type="number" value="${c.costAvoidance}" onchange="updateCountry('${uc.id}','${c.code}','costAvoidance',this.value)"><span class="vunit">k€</span></div>
        </div>
        <div class="value-cell">
          <div class="vlabel">Cost of risk</div>
          <div class="vnum"><input type="number" value="${c.costOfRisk}" onchange="updateCountry('${uc.id}','${c.code}','costOfRisk',this.value)"><span class="vunit">k€</span></div>
        </div>
        <div class="value-cell">
          <div class="vlabel">RWA savings</div>
          <div class="vnum"><input type="number" value="${c.rwaSavings}" onchange="updateCountry('${uc.id}','${c.code}','rwaSavings',this.value)"><span class="vunit">k€</span></div>
        </div>
      </div>

      <p style="color:var(--muted); font-size:11px; font-family:'IBM Plex Mono', monospace; letter-spacing:.08em; text-transform:uppercase;">
        <span class="status-dot ${statusClass}"></span>${c.status} · last update auto-saved in session
      </p>
    </div>
  `;
}

/* ===========================================================
   HELPERS
   =========================================================== */

function fieldRO(label, val, span = false) {
  return `
    <div class="field readonly ${span ? 'span-2' : ''}">
      <label>${escapeHtml(label)}</label>
      <input type="text" value="${escapeHtml(String(val))}" readonly>
    </div>`;
}
function fieldEdit(label, val, type, onchange) {
  return `
    <div class="field">
      <label>${escapeHtml(label)}</label>
      <input type="${type}" value="${escapeHtml(String(val))}" onchange="${onchange}">
    </div>`;
}
function fieldSelect(label, val, options, onchange) {
  const opts = options.map(o => `<option value="${escapeHtml(o)}"${o === val ? ' selected' : ''}>${escapeHtml(o)}</option>`).join('');
  return `
    <div class="field">
      <label>${escapeHtml(label)}</label>
      <select onchange="${onchange}">${opts}</select>
    </div>`;
}
function escapeHtml(s) {
  return String(s).replace(/[&<>"']/g, c => ({'&':'&amp;','<':'&lt;','>':'&gt;','"':'&quot;',"'":'&#39;'}[c]));
}
function truncate(s, n) { return s.length > n ? s.slice(0, n - 1) + '…' : s; }

function updateCountry(ucId, code, field, value) {
  for (const f of FLAGSHIPS) {
    const uc = f.useCases.find(u => u.id === ucId);
    if (uc) {
      const c = uc.countries.find(x => x.code === code);
      if (c) { c[field] = isNaN(value) || field === 'productionDate' || typeof c[field] === 'string' ? value : Number(value); }
    }
  }
  document.getElementById('last-updated').textContent = 'Updated · ' + new Date().toLocaleDateString('fr-FR');
}

/* ===========================================================
   MODAL — create use case, add country, export JSON
   =========================================================== */

const modal = document.getElementById('modal');

function openModal(title, bodyHtml, footerHtml) {
  document.getElementById('modal-title').textContent = title;
  document.getElementById('modal-body').innerHTML = bodyHtml;
  document.getElementById('modal-footer').innerHTML = footerHtml;
  modal.classList.add('active');
}
function closeModal() { modal.classList.remove('active'); }

function openCreateUseCase(rank) {
  openModal('New use case', `
    <div class="uc-detail-grid">
      <div class="field span-2"><label>Name</label><input type="text" id="nuc-name" placeholder="Use case name"></div>
      <div class="field span-2"><label>Description</label><textarea id="nuc-desc" placeholder="Short description"></textarea></div>
      ${fieldSelect('Pole', POLES[0], POLES, '').replace('<select', '<select id="nuc-pole"')}
      ${fieldSelect('AI Act Category', AI_ACT_CATS[0], AI_ACT_CATS, '').replace('<select', '<select id="nuc-act"')}
      ${fieldSelect('AI Domain', DOMAINS[0], DOMAINS, '').replace('<select', '<select id="nuc-domain"')}
      ${fieldSelect('AI Category', CATEGORIES[0], CATEGORIES, '').replace('<select', '<select id="nuc-cat"')}
    </div>
  `, `
    <button class="btn btn-ghost" onclick="closeModal()">Cancel</button>
    <button class="btn" onclick="createUseCase(${rank})">Create</button>
  `);
}

function createUseCase(rank) {
  const name = document.getElementById('nuc-name').value.trim();
  if (!name) { alert('Please enter a name.'); return; }
  const f = FLAGSHIPS.find(x => x.rank === rank);
  const id = `${rank}-${f.useCases.length + 1}`;
  const uc = {
    id, name,
    description: document.getElementById('nuc-desc').value.trim(),
    pole: document.getElementById('nuc-pole').value,
    entity: ENTITIES[0],
    aiActCategory: document.getElementById('nuc-act').value,
    technology: TECHS[0],
    platform: PLATFORMS[0],
    mainBrick: BRICKS[0],
    aiDomain: document.getElementById('nuc-domain').value,
    aiCategory: document.getElementById('nuc-cat').value,
    sourcing: SOURCING[0],
    iaValueTotal: 0,
    modelContributors: '',
    pilotCountry: 'France',
    itsvc2: '',
    dateItsvc: '',
    lastGate: 'Concept',
    dateLastGate: '',
    countries: [],
  };
  f.useCases.push(uc);
  closeModal();
  navigate('usecase', { flagshipId: rank, usecaseId: id });
}

function openAddCountry(ucId) {
  const f = FLAGSHIPS.find(x => x.useCases.some(u => u.id === ucId));
  const uc = f.useCases.find(u => u.id === ucId);
  const existing = new Set(uc.countries.map(c => c.code));
  const available = COUNTRIES_LIST.filter(c => !existing.has(c.code));
  if (available.length === 0) { alert('All countries already added.'); return; }
  const opts = available.map(c => `<option value="${c.code}">${c.name} (${c.code})</option>`).join('');
  openModal('Add country deployment', `
    <div class="field span-2"><label>Country</label><select id="nc-country">${opts}</select></div>
    <div class="field" style="margin-top:12px"><label>Status</label><select id="nc-status">${STATUSES.map(s => `<option>${s}</option>`).join('')}</select></div>
  `, `
    <button class="btn btn-ghost" onclick="closeModal()">Cancel</button>
    <button class="btn" onclick="addCountry('${ucId}')">Add</button>
  `);
}

function addCountry(ucId) {
  const code = document.getElementById('nc-country').value;
  const status = document.getElementById('nc-status').value;
  const country = COUNTRIES_LIST.find(c => c.code === code);
  const uc = FLAGSHIPS.flatMap(f => f.useCases).find(u => u.id === ucId);
  uc.countries.push({
    code, name: country.name,
    year: new Date().getFullYear(),
    status, productionDate: '',
    valueCapped: 0, netBanking: 0, grossCostSavings: 0,
    costAvoidance: 0, costOfRisk: 0, rwaSavings: 0,
    valueMeasureType: 'Estimated', rollout: 'Local',
    mrrCategory: 'Supporting', mrr: 0,
    useCaseOwner: '',
  });
  closeModal();
  render();
}

function openExport() {
  const payload = {
    exportedAt: new Date().toISOString(),
    flagships: FLAGSHIPS,
  };
  const json = JSON.stringify(payload, null, 2);
  openModal('Export JSON', `
    <p style="color:var(--ink-soft); font-size:13px; margin-bottom:12px;">Full registry payload — ready to push to Confluence via REST API or attach as a page property.</p>
    <div class="json-output" id="json-out">${escapeHtml(json)}</div>
  `, `
    <button class="btn btn-ghost" onclick="copyJson()">Copy</button>
    <button class="btn btn-ghost" onclick="downloadJson()">Download .json</button>
    <button class="btn" onclick="closeModal()">Done</button>
  `);
  window._exportJson = json;
}
function copyJson() {
  navigator.clipboard.writeText(window._exportJson || '');
  alert('Copied to clipboard.');
}
function downloadJson() {
  const blob = new Blob([window._exportJson || ''], { type: 'application/json' });
  const url = URL.createObjectURL(blob);
  const a = document.createElement('a');
  a.href = url;
  a.download = `flagship-registry-${new Date().toISOString().slice(0,10)}.json`;
  a.click();
  URL.revokeObjectURL(url);
}

/* close modal on overlay click */
modal.addEventListener('click', e => { if (e.target === modal) closeModal(); });
document.addEventListener('keydown', e => { if (e.key === 'Escape') closeModal(); });

/* init */
render();

</script>
</body>
</html>
```
