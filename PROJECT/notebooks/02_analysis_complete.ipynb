## Question 5 & 6: Validate earliest birth date and explain reasoning

```python
# Get detailed information about the person with earliest birth year
person_id = earliest_person.iloc[0]['nconst']
person_name = earliest_person.iloc[0]['primaryName']
birth_year = int(earliest_person.iloc[0]['birthYear'])
death_year = earliest_person.iloc[0]['deathYear']
known_titles = earliest_person.iloc[0]['knownForTitles']

print(f"Analyzing: {person_name} (ID: {person_id})")
print(f"Birth Year: {birth_year}")
print(f"Death Year: {death_year if pd.notna(death_year) else 'Not recorded'}")
print(f"\nKnown for titles: {known_titles}")

# If person has death year, calculate lifespan
if pd.notna(death_year):
    death_year_int = int(death_year)
    lifespan = death_year_int - birth_year
    print(f"\nCalculated lifespan: {lifespan} years")
    print(f"Death year: {death_year_int}")

# Look up the titles this person is known for
if pd.notna(known_titles) and known_titles != '':
    title_ids = known_titles.split(',')
    print(f"\nAnalyzing {len(title_ids)} known titles...")
    
    # Get information about these titles
    known_works = df_titles[df_titles['tconst'].isin(title_ids)]
    
    if len(known_works) > 0:
        print("\nKnown works:")
        display(known_works[['tconst', 'primaryTitle', 'titleType', 'startYear', 'endYear', 'runtimeMinutes', 'genres']])
        
        # Calculate ages at time of work
        known_works_copy = known_works.copy()
        known_works_copy['startYear'] = pd.to_numeric(known_works_copy['startYear'], errors='coerce')
        known_works_copy['age_at_work'] = known_works_copy['startYear'] - birth_year
        
        print("\nAge at time of each work:")
        for _, work in known_works_copy.iterrows():
            if pd.notna(work['age_at_work']):
                print(f"  {work['primaryTitle']} ({int(work['startYear'])}): Age {int(work['age_at_work'])}")
        
        min_age = known_works_copy['age_at_work'].min()
        max_age = known_works_copy['age_at_work'].max()
        
        if pd.notna(min_age) and pd.notna(max_age):
            print(f"\nWorking age range: {int(min_age)} to {int(max_age)} years old")
```

```python
# Additional validation: Check for statistical anomalies
print("=" * 70)
print("VALIDATION ANALYSIS")
print("=" * 70)

# 1. Check distribution of birth years
birth_year_stats = df_names['birthYear'].describe()
print("\nBirth year statistics:")
print(birth_year_stats)

# 2. Find all people born in the same century
century = (birth_year // 100) * 100
same_century = df_names[(df_names['birthYear'] >= century) & (df_names['birthYear'] < century + 100)]
print(f"\nPeople born in {century}s: {len(same_century):,}")

# 3. Check oldest 10 people
oldest_10 = df_names.nsmallest(10, 'birthYear')[['primaryName', 'birthYear', 'deathYear', 'primaryProfession']]
print("\nOldest 10 people in dataset:")
display(oldest_10)
```

### Questions 5 & 6 - Answer

**Is the birth date (year 4) correct?**

**YES, it's correct!**

**Reasoning:**


1. BirthYear fits the dataset's distribution
    Year 4 is not an isolated outlier; other very early birth years also appear in the dataset.

2. Credits are logically consistent
    The person is credited as writer for modern films. IMDb commonly lists historical authors as writers when their works are adapted—this is normal behavior in the dataset.

2. No contradiction inside IMDb data
    The titles linked to this person exist and their release years do not conflict with the ancient birth date.

**Conclusion**
Based solely on IMDb data, the birth year 4 is valid and internally consistent.

## Question 9: What is the length of the longest "short" after 1900?

```python
# Filter for "short" titles after 1900
# Convert startYear and runtimeMinutes to numeric
df_titles['startYear'] = pd.to_numeric(df_titles['startYear'], errors='coerce')
df_titles['runtimeMinutes'] = pd.to_numeric(df_titles['runtimeMinutes'], errors='coerce')

# Filter: titleType = 'short' AND startYear > 1900 AND runtimeMinutes is not null
shorts_after_1900 = df_titles[
    (df_titles['titleType'] == 'short') & 
    (df_titles['startYear'] > 1900) & 
    (df_titles['runtimeMinutes'].notna())
]

print(f"Total 'short' titles after 1900: {len(shorts_after_1900):,}")
print(f"Shorts with runtime data: {shorts_after_1900['runtimeMinutes'].notna().sum():,}")

# Find the longest short
if len(shorts_after_1900) > 0:
    longest_runtime = shorts_after_1900['runtimeMinutes'].max()
    longest_shorts = shorts_after_1900[shorts_after_1900['runtimeMinutes'] == longest_runtime]
    
    print(f"\nLength of longest 'short' after 1900: {int(longest_runtime)} minutes")
    print(f"\nTitle(s) with this runtime:")
    display(longest_shorts[['tconst', 'primaryTitle', 'startYear', 'runtimeMinutes', 'genres']])
    
    # Show statistics
    print("\nRuntime statistics for shorts after 1900:")
    print(shorts_after_1900['runtimeMinutes'].describe())
else:
    print("No shorts found after 1900 with runtime data.")
```

### Question 9 - Answer

**Length of longest 'short' after 1900:** 1311 minutes

## Question 10: What is the length of the shortest "movie" after 1900?

```python
# Filter for "movie" titles after 1900
# Filter: titleType = 'movie' AND startYear > 1900 AND runtimeMinutes is not null
movies_after_1900 = df_titles[
    (df_titles['titleType'] == 'movie') & 
    (df_titles['startYear'] > 1900) & 
    (df_titles['runtimeMinutes'].notna())
]

print(f"Total 'movie' titles after 1900: {len(movies_after_1900):,}")
print(f"Movies with runtime data: {movies_after_1900['runtimeMinutes'].notna().sum():,}")

# Find the shortest movie
if len(movies_after_1900) > 0:
    shortest_runtime = movies_after_1900['runtimeMinutes'].min()
    shortest_movies = movies_after_1900[movies_after_1900['runtimeMinutes'] == shortest_runtime]
    
    print(f"\nLength of shortest 'movie' after 1900: {int(shortest_runtime)} minutes")
    print(f"\nTitle(s) with this runtime:")
    display(shortest_movies[['tconst', 'primaryTitle', 'startYear', 'runtimeMinutes', 'genres']].head(10))
    
    # Show statistics
    print("\nRuntime statistics for movies after 1900:")
    print(movies_after_1900['runtimeMinutes'].describe())
    
    # Show distribution of very short movies
    very_short = movies_after_1900[movies_after_1900['runtimeMinutes'] < 30]
    print(f"\nNumber of movies under 30 minutes: {len(very_short):,}")
else:
    print("No movies found after 1900 with runtime data.")
```

### Question 10 - Answer

## Question 11: List of all genres represented

```python
print("Extracting all genres from dataset...\n")

# Filter out null values and split by comma
all_genres = df_titles['genres'].dropna()
genre_list = []

for genres in all_genres:
    # Split by comma and strip whitespace
    individual_genres = [g.strip() for g in str(genres).split(',')]
    genre_list.extend(individual_genres)

# Get unique genres and sort them
unique_genres = sorted(set(genre_list))

print(f"Total number of unique genres: {len(unique_genres)}")
print("\nComplete list of all genres:")
print("=" * 50)

for i, genre in enumerate(unique_genres, 1):
    print(f"{i:2d}. {genre}")

# Count frequency of each genre
print("\n" + "=" * 50)
print("Genre Frequency Analysis")
print("=" * 50)

from collections import Counter
genre_counts = Counter(genre_list)
top_10_genres = genre_counts.most_common(10)

print("\nTop 10 most common genres:")
for genre, count in top_10_genres:
    print(f"{genre:20s}: {count:,} titles")
```

### Question 11 - Answer

**list of all genres:** 28 genres

## Question 12: Highest rated comedy "movie" in the dataset

```python
# Step 1: Filter for comedy movies
print("Finding highest rated comedy movie...\n")

# Filter titles: must be 'movie' and have 'Comedy' in genres
comedy_movies = df_titles[
    (df_titles['titleType'] == 'movie') & 
    (df_titles['genres'].notna()) &
    (df_titles['genres'].str.contains('Comedy', case=False, na=False))
]

print(f"Total comedy movies: {len(comedy_movies):,}")

# Step 2: Merge with ratings data
comedy_with_ratings = comedy_movies.merge(df_ratings, on='tconst', how='inner')

print(f"Comedy movies with ratings: {len(comedy_with_ratings):,}")

# Step 3: Sort by rating (desc) and then by votes (desc) for tiebreaking
comedy_with_ratings_sorted = comedy_with_ratings.sort_values(
    by=['averageRating', 'numVotes'], 
    ascending=[False, False]
)

# Get the highest rated comedy movie
highest_rated_comedy = comedy_with_ratings_sorted.iloc[0]

print("\n" + "=" * 70)
print("HIGHEST RATED COMEDY MOVIE")
print("=" * 70)
print(f"\nTitle: {highest_rated_comedy['primaryTitle']}")
print(f"Year: {int(highest_rated_comedy['startYear']) if pd.notna(highest_rated_comedy['startYear']) else 'N/A'}")
print(f"Rating: {highest_rated_comedy['averageRating']:.1f}/10")
print(f"Number of Votes: {int(highest_rated_comedy['numVotes']):,}")
print(f"Runtime: {int(highest_rated_comedy['runtimeMinutes']) if pd.notna(highest_rated_comedy['runtimeMinutes']) else 'N/A'} minutes")
print(f"Genres: {highest_rated_comedy['genres']}")
print(f"IMDB ID: {highest_rated_comedy['tconst']}")

# Store for later use
best_comedy_id = highest_rated_comedy['tconst']
best_comedy_title = highest_rated_comedy['primaryTitle']

# Show top 10 for context
print("\n" + "=" * 70)
print("Top 10 Highest Rated Comedy Movies")
print("=" * 70)
display(comedy_with_ratings_sorted.head(10)[[
    'primaryTitle', 'startYear', 'averageRating', 'numVotes', 'genres'
]])
```

### Question 12 - Answer

**Highest rated comedy movie:** Bob vs. Society

## Question 13: Who was the director of the movie?

```python
# Find the director(s) of the highest rated comedy movie
print(f"Finding director(s) of '{best_comedy_title}'...\n")

# Look up in crew data
movie_crew = df_crew[df_crew['tconst'] == best_comedy_id]

if len(movie_crew) > 0:
    directors_ids = movie_crew.iloc[0]['directors']
    
    if pd.notna(directors_ids) and directors_ids != '':
        # Split director IDs (may be multiple)
        director_id_list = directors_ids.split(',')
        
        print(f"Director ID(s): {directors_ids}")
        print(f"Number of directors: {len(director_id_list)}\n")
        
        # Look up director names
        directors_info = df_names[df_names['nconst'].isin(director_id_list)]
        
        print("=" * 70)
        print(f"DIRECTOR(S) OF '{best_comedy_title}'")
        print("=" * 70)
        
        if len(directors_info) > 0:
            for idx, director in directors_info.iterrows():
                print(f"\nName: {director['primaryName']}")
                print(f"ID: {director['nconst']}")
                print(f"Birth Year: {int(director['birthYear']) if pd.notna(director['birthYear']) else 'N/A'}")
                print(f"Professions: {director['primaryProfession']}")
                print(f"Known for: {director['knownForTitles']}")
            
            # Display in table format
            print("\nDirector Information (Table):")
            display(directors_info[['nconst', 'primaryName', 'birthYear', 'deathYear', 'primaryProfession']])
        else:
            print("Director information not found in name.basics dataset.")
            print(f"Director ID(s): {directors_ids}")
    else:
        print("No director information available for this movie.")
else:
    print(f"Crew information not found for {best_comedy_id}")
```

### Question 13 - Answer

**Director(s) of the highest rated comedy movie:** David Pring-Mill

## Question 14: List alternate titles for the movie
